# Q1. Stop sending emails after 3 users (for loop + break)


```python
users = ["Ali", "", "Ahad", "Zara", "Mike"]
```

```python
count = 0
for user in users:
    if count == 3:
        print("Limit reached. Stop sending emails.")
        break
    print(f"Email sent to {user}")
    count += 1
```

# Q2. Stop monitoring temperature if critical reached (for loop + break)


```python
temps = [22, 25, 30, 45, 28]
critical = 40
```

```python
for t in temps:
    if t >= critical:
        print("Critical temperature reached! Stop monitoring.")
        break
    print(f"Temperature: {t}")
```

# Q3. Calculate total price of items in a shopping cart (for loop)


```python
cart = [100, 250, 50, 300]

total = 0
for price in cart:
    total += price

print(f"Total Price: {total}")
```

# Q4. Skip customers with unpaid bills (for loop + continue)


```python
customers = [
    {"name": "Ayan", "paid": True},
    {"name": "Anas", "paid": False},
    {"name": "Aliya", "paid": True}
]
```

```python
for customer in customers:
    if not customer["paid"]:
        continue
    print(f"Access granted to: {customer['name']}")
```

# Q6. Extract domain names from email list (for loop)


```python
emails = ["ahmed@gmail.com", "raza@yahoo.com", "farukh@hotmail.com"]

domains = []
for email in emails:
    domain = email.split("@")[1]
    domains.append(domain)
```

```python
print(f"domains: {domains}")
```

# Q7. Numbers divisible by 3 or 5 (List Comprehension)


```python
numbers = list(range(1, 51)) 

divisible = [n for n in numbers if n % 3 == 0 or n % 5 == 0]
```

```python
print(divisible)
```

# Q8. Multiplication table using nested for loop


```python
numbers = [2, 3, 4]
multipliers = [1, 2, 3]
```

```python
for n in numbers:
    for m in multipliers:
        print(f"{n} x {m} = {n * m}")
```

# Q9. Password prompt until correct (while loop)


```python
correct_password = "unais123"
```

```python
while True:
    password = input("Enter password: ")
    if password == correct_password:
        print("Access granted!")
        break
    else:
        print("Incorrect password. Try again.")
```

# Q10. Count down from 5 to 1, then "Time's up!" (while loop)


```python
count = 5

while count >= 1:
    print(count)
    count -= 1

print("Time's up!")
```
